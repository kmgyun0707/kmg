```python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
import time

def main():
    if not rclpy.ok():
        rclpy.init()

    node = rclpy.create_node('turtle_triangle_drawer')
    publisher = node.create_publisher(Twist, '/turtle1/cmd_vel', 10)
    
    print("노드 동작")

    try:
        for i in range(3):
            msg = Twist()
            msg.linear.x = 2.0   
            msg.angular.z = 0.0
            publisher.publish(msg)
            time.sleep(1.0)    
            
            # (잠깐 정지 - 동작 구분감을 위해)
            msg = Twist() # 0,0 상태
            publisher.publish(msg)
            time.sleep(0.5)

            # --- [2] 회전 (Turn 120도) ---
            # 120도는 라디안으로 약 2.0944 rad 입니다.
            msg = Twist()
            msg.linear.x = 0.0
            msg.angular.z = 2.0944  # 1초에 120도 돌도록 속도 설정
            publisher.publish(msg)
            time.sleep(1.0)         # 1초 동안 회전
            
            # (잠깐 정지)
            msg = Twist()
            publisher.publish(msg)
            time.sleep(0.5)
            
            print(f"{i+1}번째 변과 각도 완료")

        print("삼각형 완성!")

    except KeyboardInterrupt:
        print("사용자에 의해 중단됨")
        
    finally:
        # 3. 깔끔하게 종료
        node.destroy_node()
        rclpy.shutdown()
        print("노드 종료됨")

if __name__ == '__main__':
    main()
```
