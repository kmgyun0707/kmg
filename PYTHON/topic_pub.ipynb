```python
import rclpy as rp
from geometry_msgs.msg import Twist
rp.init()
node = rp.create_node('turtlesim_velocity_publisher')
```

```python
msg = Twist()
msg.linear.x = 2.0
msg.angular.z = 1.8
pub = node.create_publisher(Twist, '/turtle1/cmd_vel', 10)
pub.publish(msg)
```

```python
node.destroy_node()
rp.shutdown()
```
