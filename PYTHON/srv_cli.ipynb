```python
import rclpy as rp
from turtlesim.srv import TeleportAbsolute

rp.init()
node = rp.create_node('turtlesim_teleport_client')
cli = node.create_client(TeleportAbsolute, '/turtle1/teleport_absolute')
```

```python
req = TeleportAbsolute.Request()
req.x = 5.0
req.y = 5.0
req.theta = 3.14
```

```python
cli.call_async(req) 
rp.spin_once(node)
```
