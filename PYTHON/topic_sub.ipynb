```python
import rclpy as rp
from turtlebot3_msgs.srv import SetFollowPath
from rclpy.node import Node
import sys
import time
import os
import yaml
from ament_index_python.packages import get_package_share_directory
from geometry_msgs.msg import PoseStamped
from nav2_msgs.srv import LoadMap
from rclpy.qos import QoSProfile, QoSDurabilityPolicy, QoSReliabilityPolicy
from rclpy.executors import MultiThreadedExecutor
from rclpy.callback_groups import ReentrantCallbackGroup
from rclpy.task import Future
from rclpy.duration import Duration
from rclpy.time import Time
from rclpy.action import ActionClient

class PathFollowerClient(Node):
    def __init__(self):
        super().__init__('path_follower_client')
        self.cli = self.create_client(SetFollowPath, 'set_follow_path')
        while not self.cli.wait_for_service(timeout_sec=1.0):
            self.get_logger().info('service not available, waiting again...')
        self.req = SetFollowPath.Request()

    def send_request(self, path):
        self.req.path = path
        self.future = self.cli.call_async(self.req)
        rclpy.spin_until_future_complete(self, self.future)
        return self.future.result()
from nav_msgs.msg import Path
def main(args=None):
    rp.init(args=args)

    path_follower_client = PathFollowerClient()

    # Create a sample path
    path = Path()
    path.header.frame_id = "map"
    for i in range(5):
        pose = PoseStamped()
        pose.header.frame_id = "map"
        pose.pose.position.x = float(i)
        pose.pose.position.y = float(i)
        pose.pose.position.z = 0.0
        pose.pose.orientation.w = 1.0
        path.poses.append(pose)

    response = path_follower_client.send_request(path)
    path_follower_client.get_logger().info('Response: %s' % str(response.success))

    path_follower_client.destroy_node()
    rp.shutdown()
```
